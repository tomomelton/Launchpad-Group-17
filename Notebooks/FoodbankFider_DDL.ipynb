```python
import sqlite3

con = sqlite3.connect("FoodbankFinder.db")
cur = con.cursor()

cur.execute("PRAGMA foreign_keys = ON") #Required for referential integrity
```

# DDL Statements

### Foodbank

```python
cur.execute("DROP TABLE IF EXISTS Foodbank")

cur.execute("""
    CREATE TABLE Foodbank
        (
        FoodbankID INTEGER PRIMARY KEY,
        Name VARCHAR NOT NULL,
        Postcode VARCHAR NOT NULL,
        Longitude REAL NOT NULL,
        Latitude REAL NOT NULL,
        UNIQUE(Longitude, Latitude)
        )
""")
```

### Food

```python
cur.execute("DROP TABLE IF EXISTS Food")

cur.execute("""
    CREATE TABLE Food
        (
        ItemID INTEGER PRIMARY KEY,
        ItemName VARCHAR NOT NULL,
        Category VARCHAR 
        CHECK(Category IN ('Grains', 'Pasta', 'Bakery', 'Dairy', 'Protein', 'Canned Goods', 'Spreads', 'Produce'))
        )
""")
```

### FoodbankContents

```python
cur.execute("DROP TABLE IF EXISTS FoodbankContents")

cur.execute("""
    CREATE TABLE FoodbankContents
        (
        FoodbankID INTEGER,
        ItemID INTEGER,
        Category INTEGER DEFAULT 0,
        CHECK(Category IN (1, 0)),
        PRIMARY KEY(FoodbankID, ItemID),
        FOREIGN KEY(FoodbankID) 
            REFERENCES Foodbank(FoodbankID)
            ON UPDATE CASCADE
            ON DELETE CASCADE,
        FOREIGN KEY(ItemID)
            REFERENCES Food(ItemID)
            ON UPDATE CASCADE
            ON DELETE RESTRICT
        )
""")
```
